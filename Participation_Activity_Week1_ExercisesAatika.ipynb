## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
# Calculate the mean of salaries manually, as requested by the exercise.
mean_salary = sum(salaries) / len(salaries)
print(mean_salary)
```

### median

```python
median_salary = sorted(salaries)[len(salaries)//2]
print(median_salary)
```

### mode

```python
mode_salary = max(set(salaries), key=salaries.count) #import statistics mode_salary = statistics.mode(salaries)
print(mode_salary)
```

### sample variance
Remember to use Bessel's correction.

```python
variance_salary = sum ((x-mean_salary)**2 for x in salaries)/(len(salaries)-1) #import statistics variance_salary = statistics.variance(salaries)
print(variance_salary)
```

### sample standard deviation
Remember to use Bessel's correction.

```python
sd_salary = variance_salary**0.5
print (sd_salary)
```

### coefficient of variation

```python
import statistics

co_var = statistics.stdev(salaries) / statistics.mean(salaries) #co_var = sd_salary/mean_salary

print(co_var)
```

### interquartile range

```python
iqr_salary = sorted(salaries)[len(salaries)//4*3] - sorted(salaries)[len(salaries)//4*1]
print(iqr_salary)
```

### quartile coefficent of dispersion

```python
iqr_salary/mean_salary
```

## Exercise 7: Scaling data
### min-max scaling

```python
sal_scaled = [(x-min(salaries))/(max(salaries)-min(salaries)) for x in salaries]
print(sal_scaled)
```

### standardizing

```python
standardized_salaries = [(x-mean_salary)/sd_salary for x in salaries]
print(standardized_salaries)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
covariance_salary = sum((x-mean_salary)*(y-mean_salary) for x,y in zip(salaries,standardized_salaries))/(len(salaries)-1)
#import numpy as np covariance_salary = np.cov(salaries, standardized_salaries)[0, 1]

print(covariance_salary)
```

### Pearson correlation coefficient ($\rho$)

```python
from statistics import mean, stdev
mean_salary, sd_salary = mean(salaries), stdev(salaries)

standardized_salaries = [(x - mean_salary) / sd_salary for x in salaries]
standardized_salaries[:5]
import numpy as np
np.cov(sal_scaled, standardized_salaries)
```
