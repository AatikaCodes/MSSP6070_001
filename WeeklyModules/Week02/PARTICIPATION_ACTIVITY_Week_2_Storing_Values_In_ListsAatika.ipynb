3. Show the code in a new cell:

a. Create a variable called 'letters' that contains ['a',  'b',  'c', 'd'].

b. What does letters[int(int('3'*2)//11)] evaluate to? Explain why in a comment.

c. What does letters[-1] evaluate to? Explain why in a comment.

d. What does letters[:2] evaluate to? Explain why in a comment.

```python
letters = ['a', 'b', 'c', 'd'] #3.a)
print(letters)
```

```python
print(letters[int(int('3'*2)//11)])
# For 3.b)# '3'*2 gives '33', int() converts it to 33, 33//11 gives modulus of value 3,
# and letters[3] is 'd' but with print(letters[3]) gives d.
```

```python
print(letters[-1])
#For 3(c), we get the output 'd' because Python allows negative indexing to count from the end of a list.
#With positive indexing, Python starts counting from 0, beginning with the first item in the list.
#With negative indexing, Python starts counting from -1, beginning with the last item in the list.
#Therefore, letters[-1] refers to the last item, which is 'd' in this case.
```

```python
print(letters[:2])
#This is the case of 'Slicing' which is different from 'Indexing'.
#We get the output ['a', 'b'] because letters[:2] takes the items
#from the beginning of the list up to, but not including, index 2.
#Since Python starts indexing at 0, index 0 is 'a' and index 1 is 'b'.
#Therefore, the first two items, 'a' and 'b', are included in the output.
```

4. How can you get the tuple form of the value of a list? How can you get the lists from a tuple? Show a code example in a new cell in the Notebook.

```python
# Convert a list into a tuple
my_list = ['apple', 'banana', 'citrus', 'drink']
my_tuple = tuple(my_list)

print(my_tuple)

# Convert a tuple into a list
my_tuplecvt = ['apple', 'banana', 'citrus', 'drink']
my_listcvt = list(my_tuplecvt)

print(my_listcvt)
```

5. Practice program (see Sweigart, Chapter Four, Comma Code Practice Project:

a. spam = ['apples', 'bananas', 'tofu', 'cats']

Write a function that accepts the list 'spam' as an argument and returns a string with all the items separated by a comma and a space, with 'and' inserted before the last item. The function should work with any list.

```python
spam = ['apples', 'bananas', 'tofu', 'cats']
def COMMA_TEST(spam):
  if len(spam) == 0:
    return ''
  elif len(spam) == 1:
    return spam[0]
  else:
    return ', ' .join(spam[:-1])+', and ' + spam[-1]

print(COMMA_TEST(spam))
```
