**9. How can you force a variable in a function to refer to the global variable?**

Answer: Using the **Global** statement.

```python
x = 10

def change():
    global x
    x = 20

print(x)  # Output: 10
change()
print(x)  # Output: 20

#The global x statement tells Python that x refers to the global variable
#rather than creating a new local variable.
# Here, Defining the function only registered the instructions in memory without
#running them, so first print statement outputs the initial global value of 10.
#Once the function is explicitly called, its code executes and permanently
#overwrites the global variable,
#causing the final print statement to reflect the new value of 20.
```

**14. What goes in the try clause? What goes in the except clause?**

**Answer:** The **try** clause contains code that might cause an error.
The **except** clause contains code that should execute if that particular error occurs.

```python
#code that will crash without using exception handler: try and except.

num = int(input("Enter a number: "))
# If input is entered 0, the program crashes right here!
result = 100 / num

print(f"Success! The result is {result}")
```

```python
#code that will not crash using exception handler: try and except.
try:
    num = int(input("Enter a number: "))
    result = 100 / num
    print(f"Success! The result is {result}")

except ZeroDivisionError:
    print(" Error: Invalid, cannot divide a number by zero!")

except ValueError:
    print(" Error: That was not a valid whole number!")

print("The program kept running and bypassed crashing!")
```

**Answer:** Here, each function call returns the same number because the random number generator is executed only once outside the function, rather than inside it.

**The Structural Breakdown of this function is as follows:**

**The Global Variable:** When Python reads the file line-by-line, it hits line 2 (random_number = random.randint(1, 6)) and generates one single random integer (for example, 4). It saves this single value into a permanent global variable.

**The Function Behavior:** The get_random_dice_roll() function does not contain code to generate a new number. It simply reaches out, grabs the pre-calculated global random_number, and returns it.

**Repeated Calls:** No matter how many times we call get_random_dice_roll(), it will just repeat that exact same pre-saved number over and over again because random.randint is never triggered again.

```python
#fixes: If we want each function call to return a new random dice roll,
# we must move the random calculation inside the function body so that
#it runs fresh every single time the function is executed:

import random

def get_random_dice_roll():
  # Now it generates a NEW random integer every time the function is called!
    return random.randint(1, 6)

print(get_random_dice_roll())
print(get_random_dice_roll())
print(get_random_dice_roll())
```

***Part1: Making*** a function named collatz() that has one parameter named number. If number is even, then collatz() should print number // 2 and return this value. If number is odd, then collatz() should print and return 3 * number + 1.

```python
def collatz (number):
    if number % 2 == 0:
         print (number // 2, end=' ')
         return number //2
    elif number % 2 == 1:
         print (3 * number +1, end=' ')
         return 3 * number +1
```

**Now,** a program that lets the user enter an integer and that keeps calling collatz() on that number until the function returns the value 1. Also, this sequence actually works for any integer; sooner or later, using this sequence, we arrive at 1.

```python
number = int(input("Enter a number: "))
print(number, end=' ')

while (number != 1):
  number = collatz(number)
```

Here the **key operation** is:

 while (number != 1):

  number = collatz(number)

Each time through the loop:

1. collatz(number) is calculating the next number.
2. It prints that number.
3. It returns that number.
4. The returned value is stored back in number.
5. The loop continues until number becomes 1.
